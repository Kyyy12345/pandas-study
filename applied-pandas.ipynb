```python
import pandas as pd
```

```python
TITANIC_URL = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
```

```python
titanic = pd.read_csv(TITANIC_URL)
```

```python
# 등급별 agg(생존율, 요금, 인원)
titanic.groupby('Pclass').agg(생존율 = ('Survived', 'mean'), 요금 = ('Fare', 'mean'), 인원 = ('PassengerId', 'count'))
# 나이는 평균(mean), 생존 여부는 합계(sum)를 각각 구하기
titanic.groupby('Sex').agg({'Age': 'mean', 'Survived': 'sum'})
titanic.groupby('Sex')['Fare'].agg(['mean', 'min', 'max'])
```

```python
# 1등급 여성 승객 이름 조회 & 복합 조건 필터링
titanic[(titanic['Pclass'] == 1) & (titanic['Sex'] == 'female')]['Name']
```

```python
# 탑승항별 생존율 비교
titanic.groupby('Embarked')['Survived'].mean()
```
