```python
import pandas as pd
```

```python
# 데이터 생성 코드 실행 및 CSV 저장
shopping = pd.read_csv('mock_shopping_orders_with_nan.csv')
```

```python
shopping
```

```python
# price * quantity 로 매출 컬럼 생성
shopping['총매출'] = shopping['단가'] * shopping['수량']
shopping['총매출']
```

```python
#groupby + transform으로 결측치 채우기
shopping['총주문금액'] = shopping['총주문금액'].fillna(shopping['단가'] * shopping['수량'])
shopping['주문상태'] = shopping['주문상태'].fillna('확인불가')
shopping['총매출'] = shopping.groupby('카테고리')['단가'].transform(lambda x : x.fillna(x.mean()))
```

```python
shopping.isnull().sum()
```

```python
# nlargest(3) 또는 sort_values로 TOP 3 추출
shopping.sort_values('총매출', ascending=False).head(3)
```

```python
# 카테고리별 총매출 내림차순
shopping.groupby('카테고리').agg(총매출 = ('총주문금액', 'sum')).sort_values('총매출', ascending=False)
```

```python
# 고객별 총주문금액 top3
shopping.groupby('사용자ID').agg({'총주문금액': 'sum'}).sort_values('총주문금액', ascending=False).head(3)
```
