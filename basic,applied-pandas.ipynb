```python
shoppingmall_url = 'mock_shopping_orders.csv'
```

```python
#데이터 생성 코드 실행
shoppingmall = pd.read_csv(shoppingmall_url)
```

```python
# 카테고리별 총매출 내림차순
shoppingmall.groupby('카테고리').agg(총매출 = ('총주문금액', 'sum')).sort_values('총매출', ascending=False)
```

```python
shoppingmall.isnull().sum()
```

```python
# 카테고리 평균
shoppingmall['카테고리'].value_counts().mean()
```

```python
# price * quantity로 매출 컬럼 생성
shoppingmall['매출'] = shoppingmall['단가'] * shoppingmall['수량']
```

```python
# groupby + transform으로 결측치 채우기
shoppingmall['매출'] = shoppingmall.groupby('카테고리')['매출'].transform(lambda x: x.fillna(x.mean()))
```

```python
# nlargest(3) 또는 sort_values로 TOP 3 추출
shoppingmall.groupby('카테고리')['매출'].sum().nlargest(3)
#shoppingmall.groupby('카테고리')['매출'].sum().sort_values(ascending=False).head(6)
```
